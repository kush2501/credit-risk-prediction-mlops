```python
import pandas as pd
import numpy as np
```

```python
# Load Dataset.
df = pd.read_csv("../data/credit_risk.csv")
df.head()
```

```python
# Dataset shape.
print(df.shape)

# Column information.
df.info()
```

```python
# Missing values.
print(df.isnull().sum())

(df.isnull().sum() / len(df) * 100).round(2)
```

```python
# duplcated rows.
df.duplicated().sum()
```

```python
# Numericals column.
df.describe()
```

```python
# Categorical columns.
categorical_columns = df.select_dtypes(
    include="object"
).columns.tolist()

categorical_columns
```

```python
# Unique values.
for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts(dropna=False))
```

```python
# Target Distribution.
print(df["loan_status"].value_counts())

# Target Distribution in Percentage.
df["loan_status"].value_counts(normalize=True).mul(100).round(2)
```

```python
# Target + Features Relationship.

df.groupby("loan_status")[
    [
        "person_age",
        "person_income",
        "loan_amnt",
        "loan_int_rate",
        "loan_percent_income",
        "cb_person_cred_hist_length"
    ]
].mean().round(2)
```

```python
# Person Loan status whose age is greater than 100.
df[df["person_age"] > 100][
    ["person_age", "loan_status"]
]
```

```python
# Person who was employed more than 50 years.
df[df["person_emp_length"] > 50][
    ["person_age", "person_emp_length", "loan_status"]
]
```

```python
# Person age, person_income whose income is greater than 500000.
df[df["person_income"] > 500000][
    ["person_age", "person_income", "loan_status"]
]
```

```python
# Duplicated data.
df[df.duplicated(keep=False)].head(20)
```

```python
# Loan status on the basis of person_home_ownership.
pd.crosstab(
    df["person_home_ownership"],
    df["loan_status"],
    normalize="index"
).round(3)
```

```python
# Loan status on the basis of loan_intent.
pd.crosstab(
    df["loan_intent"],  
    df["loan_status"],
    normalize="index"
).round(3)
```

```python
# Loan status on the basis of loan_grade.
pd.crosstab(
    df["loan_grade"],
    df["loan_status"],
    normalize="index"
).round(3)
```

```python
# Loan status on the basis of cb_person_default_on_file.
pd.crosstab(
    df["cb_person_default_on_file"],
    df["loan_status"],
    normalize="index"
).round(3)
```
